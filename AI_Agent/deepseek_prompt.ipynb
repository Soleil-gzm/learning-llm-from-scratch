### Prompt：工程化提示词

```python
import os
import json
import requests
from dotenv import load_dotenv

# 加载.env 文件中的环境变量
load_dotenv()       # 在当前目录寻找 .env 文件

api_key = os.getenv("DEEPSEEK_API_KEY")

if not api_key:
    raise ValueError("请在 .env 文件中设置DEEPSEEK_API_KEY")

# 设置端点url
url = "https://api.deepseek.com/chat/completions"
```

#### 1.思维链

```python
# 4. 构建消息数组
# msg = [
#     # {"role":"system","content":"你是一个专业英语翻译助手，只负责将用户的内容翻译成英文，不要添加任何其他内容。"},
#     {"role":"user","content":"请用逐步推理的方式，解决一下问题：\n"
#      "如果3个苹果和4个香蕉共花费38元，2个苹果和5个香蕉共花费40元，那么单个苹果和香蕉的价格各是多少？\n​"
#      "请按照一下步骤回答：\n"
#      "1.设未知数并列出方程\n"
#      "2.解方程\n"
#      "3.验证结果\n"
#      "4.给出最终答案"
#     }
# ]

msg = [
    {"role":"system","content":"请用逐步推理的方式，解决问题：\n"
    "请按照一下步骤回答：\n"
     "1.设未知数并列出方程\n"
     "2.解方程\n"
     "3.验证结果\n"
     "4.给出最终答案"},
    {"role":"user","content":"如果3个苹果和4个香蕉共花费38元，2个苹果和5个香蕉共花费40元，那么单个苹果和香蕉的价格各是多少？"}
]

# 5. 构建请求体并转换为JSON字符串
payload = {
    "messages": msg,               # 必填字段，消息数组，包含content和role属性，至少一条消息
    "model": "deepseek-v4-flash",  # 必填字段，模型名称
    "stream": False,                # 非流式，方便一次拿到完整响应
    # "thinking": {"type": "enabled"},
    # "reasoning_effort": "high"
    "temperature":0.2
}

# 6. 设置请求头
headers = {
    "Content-Type": "application/json",         # 请求内容类型为JSON
    "Accept":'application/json',                # 接收的响应内容类型为JSON     
    "Authorization": f"Bearer {api_key}"       # API_keys,用于身份验证

}

# 7. 发送 POST 请求
response = requests.request("POST",url, headers=headers, json=payload)  # 用 json= 自动序列化，更简洁

# # 打印响应类型和内容
# print(response.text)

# 8. 处理响应
if response.status_code == 200:
    result = response.json()
    print(result)
    # 提取回复内容
    content = result["choices"][0]["message"]["content"]
    thinking = result["choices"][0]["message"]["reasoning_content"]
    print("AI 思考：", thinking)
    print("AI 回复：", content)
else:
    print(f"请求失败，状态码：{response.status_code}")
    print("错误详情：", response.text)
```
