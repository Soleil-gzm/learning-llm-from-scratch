#### 客户端获取和处理流式数据

```python
import requests
import json
from openai import OpenAI

# 设置API密钥
api_key = 'sk-18e5981052d5471ea80eba1fa8fbb08b'

# 设置DeepSeek API 的端点URL
url = "https://api.deepseek.com"

# 构建消息数组，包含系统角色提示词和用户输入
msg = [
    {"content":"You are a heplful assistant","role":"system"},       # 系统角色提示词
    {"content":"宋朝延续多少年，输出不超过50字","role":"user"}           # 用户输入的消息
]

# 构建请求体并转换为Json字符串
payload = json.dumps({
    "message": msg,             # 必填字段，消息数组，包含content和role属性，至少一条消息
    "model":"deepseek-v4-pro",    # 必填字段，模型名称，可选"deepseek-chat"或"deepseek-reasoner"
    "stream":True,              # 启用流式输入
    "max_tokens": 60            # 设置最大生成token数

})

# 设置HTTP请求头，包含身份验证和内容类型
headers = {
    'Content-Type':'application/json',           # 请求内容类型为JSON
    'Accept':'application/json',                 # 接收的响应内容为JSON
    'Authorization': f'Bearer {api_key}'         # API keys, 用于身份验证 
}

# 逐块处理响应
try:
    response = requests.request("POST",url,headers=headers,data=payload,stream=True)    # 发送POST请求并获取响应
    response.raise_for_status()             # 检查HTTP响应状态，如果不是200则抛出异常

    for chunk in response.iter_lines():     # 逐行读取流式响应
        if chunk:   # 如果数据块不为空
            decode_chunk = chunk.decode('utf-8').strip()    # 将字节流解码为字符串并去除首尾空白

            if decode_chunk.startswith("data:"):            # 判断是否为数据行
                data_str =decode_chunk[6:]                  # 提取数据内容（去掉前缀"data:")
                if data_str =="[DONE]":             # 流结束标志
                    break                           # 终止循环，结束读取

                try:
                    json_data =json.loads(data_str) # 解析Json数据
                    if "choices" in json_data:
                        content =json_data["choices"][0].get("delta",{}).get("content","")
                        print(content,end="",flush=True)        # 逐字输出，刷新缓冲

                except json.JSONDecodeError:
                    # 解析失败时忽略，继续处理下一块数据
                    continue


except requests.exceptions.RequestException as e:
    print(f"\n请求失败：{e}")       # 打印错误信息


```
